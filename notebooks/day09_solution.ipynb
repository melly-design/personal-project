```python
import pandas as pd
import numpy as np
df = pd.read_csv('aice_practice.csv')
```

## 1. [빈칸] sum 두 번 = 전체 합

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (sum 두 번 = 전체 합):
#   df.isnull().sum().sum()
int(df.isnull().sum().sum())
```

## 2. [빈칸] subset = 이 컬럼만 보고

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (subset = 이 컬럼만 보고):
#   df.dropna(subset=['Age']).shape
df.dropna(subset=['Age']).shape
```

## 3. [오류 정정] 함수는 괄호로 실행해야 값이 나온다

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (함수는 괄호로 실행해야 값이 나온다):
#   df['Age'] = df['Age'].fillna(df['Age'].median())
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Age'].mean()
```

## 4. [빈칸] q1 = 0.25, q3 = 0.75

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (q1 = 0.25, q3 = 0.75):
#   q3 = df['Fare'].quantile(0.75)
round(df['Fare'].quantile(0.75), 2)
```

## 5. [출력 예측] 고른 분포 → 이상치 적음

```python
df = pd.read_csv('aice_practice.csv')
# 결과: 1명   (고른 분포 → 이상치 적음)
q1 = df['Age'].quantile(0.25)
q3 = df['Age'].quantile(0.75)
iqr = q3 - q1
low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
int(((df['Age'] < low) | (df['Age'] > high)).sum())
```

## 6. [빈칸] axis=0 행, axis=1 컬럼

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (axis=0 행, axis=1 컬럼):
#   df_temp = df.drop(df[cond].index, axis=0)
cond = df['Fare'] > 300
len(df.drop(df[cond].index, axis=0))
```

## 7. [출력 예측] 담거나 inplace=True

```python
df = pd.read_csv('aice_practice.csv')
# 결과: 599   (담거나 inplace=True)
df_temp = df.drop(df[df['Fare'] > 300].index, axis=0)
df_temp.reset_index(drop=True)
int(df_temp.index[-1])
```

## 8. [빈칸] 여러 값 중 하나 = isin

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (여러 값 중 하나 = isin):
#   df.loc[df['Embarked'].isin(['C', 'Q']), ['Embarked']] = 'other'
df.loc[df['Embarked'].isin(['C', 'Q']), ['Embarked']] = 'other'
int((df['Embarked'] == 'other').sum())
```

## 9. [빈칸] replace = 짝대로 바꾸기, 없는 값은 그대로

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (replace = 짝대로 바꾸기, 없는 값은 그대로):
#   df['Sex'] = df['Sex'].replace({'male': 0, 'female': 1})
df['Sex'] = df['Sex'].replace({'male': 0, 'female': 1})
int(df['Sex'].sum())
```

## 10. [오류 정정] 옵션 이름도 철자까지

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (옵션 이름도 철자까지):
#   pd.get_dummies(df, columns=['Sex'])
pd.get_dummies(df, columns=['Sex'])
```

## 11. [빈칸] (추가) 글자 결측은 mode()[0]

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (글자 결측은 mode()[0]):
#   df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
int(df['Embarked'].isnull().sum())
```

## 12. [오류 정정] (추가) 글자 컬럼이 섞이면 numeric_only=True, 2편 corr 와 같은 이유. Age 만 골라 채워도 정답

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (글자 컬럼이 섞이면 numeric_only=True, 2편 corr 와 같은 이유. Age 만 골라 채워도 정답):
#   df = df.fillna(df.median(numeric_only=True))
df = df.fillna(df.median(numeric_only=True))
int(df['Age'].isnull().sum())
```

## 13. [출력 예측] (추가) dropna 는 빈칸이 하나라도 있는 행을 지운다

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (dropna 는 빈칸이 하나라도 있는 행을 지운다):
#   480
len(df.dropna())
```

## 14. [빈칸] (추가) 펼칠 컬럼 목록 = columns (Embarked 1개 → C·Q·S 3개)

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (펼칠 컬럼 목록 = columns (Embarked 1개 → C·Q·S 3개)):
#   pd.get_dummies(data=df, columns=['Embarked']).shape[1]
pd.get_dummies(data=df, columns=['Embarked']).shape[1]
```

## 15. [출력 예측] (추가) map 은 사전에 없는 값을 NaN 으로 (replace 와 다른 점)

```python
df = pd.read_csv('aice_practice.csv')
# 정답 (map 은 사전에 없는 값을 NaN 으로 (replace 와 다른 점)):
#   226
int(df['Sex'].map({'male': 0}).isnull().sum())
```
